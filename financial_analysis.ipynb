## Financial Analysis: Microsoft, Tesla, and Apple (2023–2025)

### Introduction

This report analyzes the financial performance of Microsoft, Tesla, and Apple from fiscal years 2023 to 2025. The analysis focuses on five financial metrics: total revenue, net income, total assets, total liabilities, and operating cash flow. Financial data were extracted from the companies' annual Form 10-K filings and analyzed using Python and the pandas library to calculate year-over-year changes and compare financial trends.

### Data Loading

The financial dataset was imported from a CSV file using the pandas library.

```python
import pandas as pd

df = pd.read_csv("k10forge.csv")
df.head()
```

### Data Validation

Before performing financial calculations, the dataset was inspected to verify its structure, data types, and missing values.

```python
df.info()
```

### Financial Calculations

The following calculations examine year-over-year financial changes, profitability, and the relationship between total liabilities and total assets.

#### Revenue Growth

Revenue growth measures the percentage change in revenue compared with the previous fiscal year.

$$
\text{Revenue Growth} = \frac{\text{Current Year Revenue} - \text{Previous Year Revenue}}{\text{Previous Year Revenue}} \times 100\%
$$

```python
df = df.sort_values(["company", "year"])

df["revenue growth (%)"] = (
    df.groupby("company")["total revenue"].pct_change() * 100
)

df[["company", "year", "total revenue", "revenue growth (%)"]]
```

#### Net Income Growth

Net income growth measures the percentage change in a company's net income compared with the previous fiscal year.

$$
\text{Net Income Growth} = \frac{\text{Current Year Net Income} - \text{Previous Year Net Income}}{\text{Previous Year Net Income}} \times 100\%
$$

```python
df["net income growth (%)"] = (
    df.groupby("company")["net income"].pct_change() * 100
)

df[["company", "year", "net income", "net income growth (%)"]]
```

#### Net Profit Margin

Net profit margin measures the percentage of revenue that remains as net income after expenses.

$$
\text{Net Profit Margin} = \frac{\text{Net Income}}{\text{Total Revenue}} \times 100\%
$$

```python
df["profit margin (%)"] = df["net income"] / df["total revenue"] * 100

df[["company", "year", "profit margin (%)"]]
```

#### Operating Cash Flow Growth

Operating cash flow growth measures the percentage change in cash generated from operating activities compared with the previous fiscal year.

$$
\text{Operating Cash Flow Growth} =
\frac{\text{Current Year Operating Cash Flow} - \text{Previous Year Operating Cash Flow}}
{\text{Previous Year Operating Cash Flow}}
\times 100\%
$$

```python
df["cash flow growth (%)"] = (
    df.groupby("company")["operating cash flow"].pct_change() * 100
)

df[["company", "year", "operating cash flow", "cash flow growth (%)"]]
```

#### Liabilities-to-Assets Ratio

The liabilities-to-assets ratio measures the proportion of a company's total assets that corresponds to its total liabilities.

$$
\text{Liabilities-to-Assets Ratio} =
\frac{\text{Total Liabilities}}{\text{Total Assets}}
\times 100\%
$$

```python
df["liabilities to assets (%)"] = (
    df["total liabilities"] / df["total assets"] * 100
)
df[["company", "year","total assets", "liabilities to assets (%)"]]
```

### Company Analysis

### Tesla: Financial Performance

From 2023 to 2025, Tesla experienced a significant decline in profitability. Revenue growth slowed and eventually turned negative, while its net profit margin dropped from 15.47% to 4.07%. However, its liabilities-to-assets ratio remained relatively stable at around 40%. These figures alone are insufficient to determine whether Tesla was experiencing serious financial difficulties.

### Microsoft: Financial Performance

Microsoft demonstrated consistent financial growth from 2023 to 2025. The increases in revenue, net income, and profit margin suggest that the company experienced business expansion while improving its profitability. However, these financial indicators alone are insufficient to determine which specific strategies contributed to this growth.

### Apple: Financial Performance

Apple demonstrated a decline followed by a recovery in profitability from 2023 to 2025. In 2024, revenue increased while the net profit margin declined. In 2025, however, revenue, net income, and net profit margin all increased, indicating a recovery in profitability. Nevertheless, the underlying causes of these changes cannot be determined from these financial figures alone.

### Comparative Conclusion

Overall, Microsoft, Tesla, and Apple demonstrated different financial trends from 2023 to 2025. Microsoft maintained consistent growth, Apple experienced a recovery in profitability after an initial decline, while Tesla experienced a significant decline in profitability. These differences highlight the importance of evaluating financial performance using multiple indicators, including revenue, net income, operating cash flow, and financial structure, rather than relying on a single metric.
